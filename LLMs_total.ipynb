**What is an LLM? — Next Token Prediction Demo**

```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="distilgpt2"
)

prompt = "Tomorrow the weather will be"

result = generator(
    prompt,
    max_new_tokens=4,
    num_return_sequences=3
)

for i, output in enumerate(result, 1):
    print(f"\nOutput {i}:")
    print(output["generated_text"])
```

**Traditional Programming vs LLM**

```python
def rule_based_reply(user_input):
    user_input = user_input.lower()

    if "hello" in user_input:
        return "Hi!"
    elif "ticket" in user_input:
        return "Please check the official ticket price."
    elif "train" in user_input:
        return "Please check the train schedule for the latest timing."
    else:
        return "Sorry, I don't understand."

print(rule_based_reply("hello"))
print(rule_based_reply("What time is the train?"))
print(rule_based_reply("Can you suggest a place to visit?"))
```

**Tokens**

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("gpt2")

text = "The weather in Hyderabad is sunny today!"

tokens = tokenizer.tokenize(text)
token_ids = tokenizer.encode(text)

print("Original Text:")
print(text)

print("\nTokens:")
print(tokens)

print("\nToken IDs:")
print(token_ids)

print("\nNumber of Tokens:")
print(len(token_ids))
```

```python
examples = [
    "The weather is sunny",
    "WeatherForecasting",
    "Travel Planning",
    "Hyderabad Railway Station",
]

for text in examples:
    tokens = tokenizer.tokenize(text)
    token_ids = tokenizer.encode(text)

    print("\nText:", text)
    print("Tokens:", tokens)
    print("Token Count:", len(token_ids))
```

**Embeddings**

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

sentences = [
    "I enjoy travelling by train",
    "Taking the train is my favorite way to travel",
    "The weather is very cold"
]

embeddings = model.encode(sentences)

for sentence, vector in zip(sentences, embeddings):
    print("\nSentence:", sentence)
    print("Embedding size:", len(vector))
    print("First 5 numbers:", vector[:5])
```

**Why Embeddings Matter? — Similarity Demo**

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

sentences = [
    "I enjoy travelling by train",
    "Taking the train is my favorite way to travel",
    "The weather is very cold",
    "Hyderabad has many popular tourist places"
]

embeddings = model.encode(sentences)

query = "I want to travel by train"

query_embedding = model.encode([query])

scores = cosine_similarity(
    query_embedding,
    embeddings
)[0]

for sentence, score in zip(sentences, scores):
    print(f"{sentence} --> Similarity: {score:.3f}")
```

**Context Window Demo**

```python
text = """
A travel assistant helps people plan trips and find useful information.
It can compare destinations, suggest routes, and summarize travel options.
Travelers should verify schedules and prices with official sources before booking.
"""
len(text.split())
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("gpt2")

context_limit = 50

text = """
A travel assistant helps people plan trips and find useful information.
It can compare destinations, suggest routes, and summarize travel options.
Travelers should verify schedules and prices with official sources before booking.
"""

tokens = tokenizer.encode(text)

print("Total Tokens:", len(tokens))
print("Context Limit:", context_limit)

if len(tokens) > context_limit:
    print("Text is too long for the context window.")
    allowed_tokens = tokens[:context_limit]
    shortened_text = tokenizer.decode(allowed_tokens)

    print("\nShortened Text:")
    print(shortened_text)
else:
    print("Text fits inside the context window.")
```
